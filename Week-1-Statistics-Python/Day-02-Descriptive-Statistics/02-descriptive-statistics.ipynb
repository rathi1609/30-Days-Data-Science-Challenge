# Importing 

```python
import pandas as pd
```

# Create a small dataset

```python
df = pd.DataFrame({
    "Daily_Earnings": [500, 600, 650, 700, 5000]
})
```

```python
print("Mean:", df["Daily_Earnings"].mean())
print("Median:", df["Daily_Earnings"].median())
print("Mode:", df["Daily_Earnings"].mode().tolist())
print("Variance:", df["Daily_Earnings"].var())
print("Standard Deviation:", df["Daily_Earnings"].std())
```

# Summary: quartiles, min, max, etc.

```python
print("\nSummary Statistics:")
print(df["Daily_Earnings"].describe())
```

# Calculate quartiles separately

```python
print("\nQuartiles:")
print(df["Daily_Earnings"].quantile([0.25, 0.50, 0.75]))
```
