# Import libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# Understand basic probability
\[P(A)=\frac{\text{Favourable outcomes}}{\text{Total equally likely outcomes}}\]

```python
# Example

favourable_outcomes = 1
total_outcomes = 2

probability = favourable_outcomes / total_outcomes

print("Probability of heads:", probability)
print("Probability percentage:", probability * 100, "%")
```

```python
np.random.seed(42)

# Generate 1,000 coin tosses: 1 = heads, 0 = tails
tosses = np.random.choice([0, 1], size=1000)

heads = np.sum(tosses == 1)
tails = np.sum(tosses == 0)

print("Total tosses:", len(tosses))
print("Heads:", heads)
print("Tails:", tails)
print("Experimental probability of heads:", heads / len(tosses))
```

# Normal distribution

```python
# Generate 1,000 observations
data = np.random.normal(
    loc=50,     # Mean
    scale=10,   # Standard deviation
    size=1000   # Number of observations
)

print("First 10 values:")
print(data[:10])

print("Mean:", round(np.mean(data), 2))
print("Standard deviation:", round(np.std(data), 2))
```

```python
# Visualize the normal distribution
plt.figure(figsize=(8, 5))

plt.hist(data, bins=30, edgecolor="black", alpha=0.7)
plt.axvline(np.mean(data), linestyle="--", label="Mean")

plt.title("Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
mean = np.mean(data)
std = np.std(data)

within_1_std = np.mean(
    (data >= mean - std) & (data <= mean + std)
)

within_2_std = np.mean(
    (data >= mean - 2 * std) & (data <= mean + 2 * std)
)

print("Within 1 standard deviation:",
      round(within_1_std * 100, 2), "%")

print("Within 2 standard deviations:",
      round(within_2_std * 100, 2), "%")
```

# Binomial distribution

```python
# Simulate 1,000 experiments
# Each experiment contains 10 coin tosses
data = np.random.binomial(n=10, p=0.5, size=1000)

print("First 10 experiments:")
print(data[:10])

print("Average number of heads:",
      round(np.mean(data), 2))
```

```python
# Count how often each number of heads occurs
values, counts = np.unique(data, return_counts=True)
```

```python
# Visualize the binomial distributionvalues, counts = np.unique(data, return_counts=True)

plt.figure(figsize=(8, 5))
plt.bar(values, counts, edgecolor="black")

plt.title("Binomial Distribution")
plt.xlabel("Number of Heads in 10 Tosses")
plt.ylabel("Frequency")
plt.xticks(range(11))
plt.grid(axis="y", alpha=0.3)

plt.show()
```

# Skewed distribution

```python
# Generate right-skewed data
data = np.random.exponential(scale=2, size=1000)

print("Mean:", round(np.mean(data), 2))
print("Median:", round(np.median(data), 2))

plt.figure(figsize=(8, 5))
plt.hist(data, bins=30, edgecolor="black")
plt.axvline(np.mean(data), linestyle="--", label="Mean")

plt.axvline(np.median(data), linestyle=":", label="Median")

plt.title("Right-Skewed Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```
