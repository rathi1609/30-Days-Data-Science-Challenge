```python
from scipy.stats import ttest_1samp
```

```python
# Scores of 10 students after a new teaching method
scores = [78, 82, 85, 80, 88, 84, 79, 86, 83, 87]
```

```python
# Scores of 10 students after a new teaching method
scores = [78, 82, 85, 80, 88, 84, 79, 86, 83, 87]
```

```python
# Variables define karna zaroori hai
historical_average = 75
alpha = 0.05
```

```python
# One-sample t-test
t_stat, p_value = ttest_1samp(
    scores,
    popmean=historical_average,
    alternative="greater"
)
```

```python
print("Average score:", sum(scores) / len(scores))
print("T-statistic:", round(t_stat, 3))
print("P-value:", round(p_value, 5))

if p_value < alpha:
    print("Reject H0: Evidence that the average score is higher.")
else:
    print("Fail to reject H0: Insufficient evidence of a higher average.")
```
