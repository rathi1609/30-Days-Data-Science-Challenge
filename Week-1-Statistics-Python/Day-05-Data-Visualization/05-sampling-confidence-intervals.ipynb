# Import libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

# Create a population dataset

```python
np.random.seed(42)

population = np.random.normal(
    loc=100,
    scale=15,
    size=10000
)

print("Population size:", len(population))
print("Population mean:", round(np.mean(population), 2))
```

# Select one random sample

```python
sample = np.random.choice(population, size=50, replace=False)

print("Sample size:", len(sample))
print("Sample mean:", round(np.mean(sample), 2))
```

# Compare sample means across repeated samples

```python
sample_means = []

for _ in range(1000):
    sample = np.random.choice(population, size=50, replace=False)
    sample_means.append(np.mean(sample))

sample_means = np.array(sample_means)

print("Average sample mean:", round(np.mean(sample_means), 2))
```
